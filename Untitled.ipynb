```python
import pandas as pd
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline, TrainingArguments, Trainer
import torch
from peft import LoraConfig, get_peft_model, TaskType
```

```python
model = "Qwen/Qwen3-0.6B"
pipeline = pipeline("text-generation",
                    model=model,
                    device='cuda')
```

```python
prompt = "hi"
answer = pipeline(prompt)
print(answer)
```

```python
df = pd.read_json('data.json')
df.head()
```

```python
tokenizer = AutoTokenizer.from_pretrained(model)
```

```python
def preprocess_function(row):
  text = row['problem'] + '/n' + row['solution']
  tokenized_text = tokenizer(text,
                           max_length=128,
                           truncation=True,
                           padding='max_length',
                           return_tensors=None)
  tokenized_text['labels']=tokenized_text['input_ids'].copy()
  return tokenized_text
```

```python
processed_data = df.apply(preprocess_function, axis=1)
```

```python
processed_data
```

```python
tokenized_encoding = {
    "input_ids": [x["input_ids"] for x in processed_data],
    "attention_mask": [x["attention_mask"] for x in processed_data],
    "labels": [x["labels"] for x in processed_data]
}
```

```python
autoModel = AutoModelForCausalLM.from_pretrained(
    model,
    device_map="cuda",
    dtype=torch.float16,
    )
```

```python
lora_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    target_modules=["q_proj", "k_proj", "v_proj"]
)
```

```python
peftModel = get_peft_model(autoModel, lora_config)
```

```python
trainin_args = TrainingArguments(
    num_train_epochs=35,
    learning_rate=0.001
)
```

```python
class CustomDataset(torch.utils.data.Dataset):
    def __init__(self, encodings):
        self.encodings = encodings

    def __getitem__(self, idx):
        return {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}

    def __len__(self):
        return len(self.encodings["input_ids"])
```

```python
train_dataset = CustomDataset(tokenized_encoding)
```

```python
trainer = Trainer(
    model=peftModel,
    args=trainin_args,
    train_dataset=train_dataset,
)
```

```python
trainer.train()
```

```python
trainer.save_model('./first_model')
tokenizer.save_pretrained('./first_model')
```

```python
from transformers import pipeline

new_pipeline_instance = pipeline(
    "text-generation",
    model="./first_model",
    tokenizer="./first_model",
    device='cuda'
)

prompt = "رمز عبورم را یادم رفته چیکار کنم؟"
answer = new_pipeline_instance(prompt)
print(answer)
```
